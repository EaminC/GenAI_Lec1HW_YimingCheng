Import required libraries

```python
print(sys.path)
import torch
from pytorch_transformers import GPT2Tokenizer, GPT2LMHeadModel
```

Load pre-trained model tokenizer (vocabulary)

```python
tokenizer = GPT2Tokenizer.from_pretrained('gpt2')
```

Encode the sentence to complete

```python
text = "The answer to the universe is not to be found in the"
indexed_tokens = tokenizer.encode(text)
```

Convert indexed tokens in a PyTorch tensor

```python
tokens_tensor = torch.tensor([indexed_tokens])
```

Load pre-trained model (weights)

```python
model = GPT2LMHeadModel.from_pretrained('gpt2')
```

Set the model in evaluation mode to deactivate the DropOut modules

```python
model.eval()
```

If you have a GPU, uncomment the following to put everything on cuda

```python
# tokens_tensor = tokens_tensor.to('cuda')
# model.to('cuda')

tokens_tensor = tokens_tensor.to('mps')
model.to('mps')
```

Predict all tokens

```python
with torch.no_grad():
    outputs = model(tokens_tensor)
    predictions = outputs[0]
```

Get the predicted next sub-word

```python
predicted_index = torch.argmax(predictions[0, -1, :]).item()
predicted_text = tokenizer.decode(indexed_tokens + [predicted_index])
```

Print the predicted word

```python
print(predicted_text)
```

```python
# Extract probabilities for the next token
next_token_logits = predictions[0, -1, :]
probabilities = torch.softmax(next_token_logits, dim=-1)

# Decode all tokens and probabilities
top_k = 10  # Number of top tokens to display
sorted_indices = torch.argsort(probabilities, descending=True)

# Print the top tokens and their probabilities
print("Top tokens excluding 'universe':")
for idx in sorted_indices[:top_k]:
    token = tokenizer.decode([idx.item()]).strip()
    if token != "universe":  # Exclude the token "universe"
        print(f"Token: '{token}', Probability: {probabilities[idx]:.4f}")
```

```python
# Generate text of varying lengths
device = torch.device("mps")
# Function to generate text with a specified number of tokens
def generate_text(input_text, num_tokens, model, tokenizer, device):
    indexed_tokens = tokenizer.encode(input_text)
    tokens_tensor = torch.tensor([indexed_tokens]).to(device)
    
    model.eval()
    with torch.no_grad():
        for _ in range(num_tokens):
            outputs = model(tokens_tensor)
            predictions = outputs[0]
            predicted_index = torch.argmax(predictions[0, -1, :]).item()
            indexed_tokens.append(predicted_index)
            tokens_tensor = torch.tensor([indexed_tokens]).to(device)
    
    return tokenizer.decode(indexed_tokens)

# Input text
input_text = "The answer to the universe is not to be found in the"

# Generate texts of different lengths
lengths = [1, 5, 10, 20]
for length in lengths:
    generated_text = generate_text(input_text, length, model, tokenizer, device)
    print(f"\nGenerated text ({length} tokens):\n{generated_text}")
```

```python
# Generate text for a single role and query
role = "scientist"
query = "explain the concept of time."

# Combine role and query
input_text = f"As a {role}, {query}"
indexed_tokens = tokenizer.encode(input_text)
tokens_tensor = torch.tensor([indexed_tokens]).to(device)

# Generate text
num_tokens = 20
with torch.no_grad():
    for _ in range(num_tokens):
        outputs = model(tokens_tensor)
        predictions = outputs[0]
        predicted_index = torch.argmax(predictions[0, -1, :]).item()
        indexed_tokens.append(predicted_index)
        tokens_tensor = torch.tensor([indexed_tokens]).to(device)

# Decode and print
generated_text = tokenizer.decode(indexed_tokens)
print(f"Role: {role}, Query: {query}\nGenerated Text:\n{generated_text}")
```
